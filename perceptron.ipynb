```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df  = pd.read_csv("data.csv")
# df
# X = df[df['Class'].drop(axis=1)].copy()

X1 = df['Feature1']
X2 = df['Feature2']


df['new_feature1'] =( (df['Feature1']- df['Feature1'].mean())/ df['Feature1'].std())
df['new_feature2'] =( (df['Feature2']- df['Feature2'].mean())/ df['Feature2'].std())

M1 = df['new_feature1']
M2 = df['new_feature2']
colors = np.where(df['Class'] == 0, 'green', 'orange')

plt.scatter(M1, M2, c=colors, s=80)
plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.show()
```

```python
# Input matrix (n_samples, n_features)
X = np.array(df[['new_feature1', 'new_feature2']].values)

# Output labels
y = df['Class'].values

from random import randint
```

```python
def perceptron(X,y):
    X = np.insert(X, 0, 1, axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    epochs = 1000

    for i in range(epochs):
        j = np.random.randint(0,65)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]

    return weights[0],weights[1:]

```

```python
def step(z):
    return 1 if z>0 else 0
```

```python
intercept_,coef_ = perceptron(X,y)
```

```python
print(coef_)
print(intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b =   -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
plt.plot(x_input,y_input,'r')
plt.scatter(M1, M2, c=colors, s=20)
plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.show()
```
