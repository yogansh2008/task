```python
# Detect missing values
import pandas as pd
import numpy as np
fl = pd.read_csv('tios2.csv')
print(fl.head())
print(fl.isnull().sum())
```

```python
# Handle missing values using:
#
# Drop rows
#
# Fill with mean / median / mode
import pandas as pd
import numpy as np
fl = pd.read_csv('tios2.csv')

fl['total_bill'] =fl['total_bill'].fillna(fl['total_bill'].mean())
fl['tip']=fl['tip'].fillna(fl['tip'].mode().iloc[0])

fl['time'].dropna(inplace=True)
fl['day'].dropna(inplace=True)
fl.to_csv('cleaned_tios2.csv', index=False)

```

```python
import pandas as pd
import numpy as np
df = pd.read_csv('tips.csv')
# print(df.head())
df['Tip_percent'] = df['tip'] / df['total_bill'] * 100
print(df.head())

# Create categorical labels:
# Low / Medium / High bill

df['bill_category'] = pd.cut(df['total_bill'], bins = [0,15,30,df['total_bill'].max()],labels =[ 'low', 'medium', 'high'])
print(df.head())
```

```python
# Group data by:
# Day
# Time
# Sex
# Smoker
# Calculate:
# Mean tip
# Total bill sum
# Count of customer

import pandas as pd
import numpy as np
df = pd.read_csv('tips.csv')
df.groupby('day')['total_bill'].mean()
df.groupby('time')['total_bill'].mean()
df.groupby(['time','day'])['total_bill'].mean()
df.groupby('time')['tip'].mean()
df.groupby('sex').count()
# Perform multiple aggregations at once

df.groupby('day').agg({
    'total_bill': ['mean', 'max', 'min'],
    'tip': ['mean', 'sum']
})
```
